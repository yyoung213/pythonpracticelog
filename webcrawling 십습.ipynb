```python
#멜론론
import requests
from bs4 import BeautifulSoup

headers={'User-Agent' :'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36'}
data = requests.get('https://www.melon.com/chart/index.htm',headers=headers)
soup = BeautifulSoup(data.text, 'html.parser') #html 받아오기기
#print(soup)
title = soup.select_one('#lst50 > td:nth-child(6) > div > div > div.ellipsis.rank01 > span > a') #1등 제목 한번 뽑아와봤음음
#print(title.text)

# #lst50 > td:nth-child(2) > div > span.rank  1위
# #lst50 > td:nth-child(6) > div > div > div.ellipsis.rank01 > span > a   1위 제목
# #lst50 > td:nth-child(6) > div > div > div.ellipsis.rank02 > a         1위 가수
# #lst50 > td:nth-child(2) > div > span.rank  2위
# #lst50 > td:nth-child(6) > div > div > div.ellipsis.rank01 > span > a    2위 제목목
# #lst50 > td:nth-child(6) > div > div > div.ellipsis.rank02 > a           2위 가수수
top_100 = soup.select('#lst50 > td > div') #tr로 이루어진 list # top 100이라는 변수에 soup html을 조금 간추려서 넣음
#print(top_100)

for tr in top_100 : #html 한줄씩 들어감감
    rank = tr.select_one('span.rank') #하 ㄴ줄씩 들어간 html 에 span.rank 잇으면 rank 된는 거고 아니면 none
        
    if rank is not None: 
        print(rank.text, end=" ") #1
        
    title = tr.select_one('div > div.ellipsis.rank01 > span > a') #제목목
    
    if title is not None:
        print(title.text)# Too Bad

    singer = tr.select_one('div > div.ellipsis.rank02 > a ')
    
    if singer is not None:
        print(singer.text)# GD
```
