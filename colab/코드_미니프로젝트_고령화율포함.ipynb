
서울시 구별 한방기관 수와 평균소득 상관분석 (3가지 버전)

① 서울시 구별 한방병원 수  vs 소득

② 인구 10만명당 한방기관 수 vs 소득

① 서울시 구별 한방의료기관 수 Vs 평균소득

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
institution = pd.read_csv("/content/sample_data/institution.csv")
income = pd.read_csv("/content/sample_data/income.csv")
institution_per_10 = pd.read_csv("/content/sample_data/institution_per_10.csv")
```

```python
df = institution.merge(income, on='구').merge(institution_per_10, on='구')
df.rename(columns={'기관수': "한방기관"}, inplace=True)
df
```

```python
df_ex = df[~df['구'].isin(['종로구', '중구'])].copy() #종로, 중구 제외
```

```python
# 상관관계 계산
r1 = df['한방기관'].corr(df['평균소득'])
r2 = df['한방기관_10만명당'].corr(df['평균소득'])
r3 = df_ex['한방기관_10만명당'].corr(df_ex['평균소득'])

print(r1,r2,r3)
```

```python
plt.figure()
plt.scatter(df['평균소득'], df['한방기관'])
m1, b1 = np.polyfit(df['평균소득'], df['한방기관'], 1)
x1 = np.array([df['평균소득'].min(), df['평균소득'].max()])
plt.plot(x1, m1 * x1 + b1, linestyle='--')
plt.title(f"① 절대 기관수 vs 평균소득 (r = {r1:.2f})")
plt.xlabel("평균소득 (백만원)")
plt.ylabel("한방기관 수")
plt.tight_layout()
plt.show()
```

# 피어슨 상관계수 r ≈ 0.58 → 통계적으로 유의미한 중간 정도의 양(+)의 상관.

# 근로소득 평균이 높은 구일수록 한방의료기관(한의원 + 한방병원) 개수가 많아지는 경향이 확인

소득 수준 ↔ 수요

고소득 지역일수록 건강 수요(한방진료·치료·미용 포함)가 크고, 지불 여력도 높음

한방의료기관은 상대적으로 자비(보험 비급여)비중이 큰 편이라 민간 소비 수준과 밀접한 연관

입지 경쟁 & 상권

강남·서초·송파처럼 상업·의료가 발달한 구는 의료기관 전반이 밀집해 있어 신규 개원이 활발합니다.

데이터의 한계

```python
# [2] 인구 10만명당 한병병원 수 vs 평균소득
plt.figure()
plt.scatter(df['평균소득'], df['한방기관_10만명당'])
m2, b2 = np.polyfit(df['평균소득'], df['한방기관_10만명당'], 1)
x2 = np.array([df['평균소득'].min(), df['평균소득'].max()])
plt.plot(x2, m2 * x2 + b2, linestyle='--')
plt.title(f"[②] 10만명당 기관수 vs 평균소득 (r = {r2:.2f})")
plt.xlabel("평균소득 (백만원)")
plt.ylabel("한방기관 수 (10만명당)")
plt.tight_layout()
plt.show()

```

```python
import seaborn as sns
import scipy.stats as stats
```

```python
inst10 = pd.read_csv('/content/sample_data/institution_per_10.csv')   # 구, 한방기관_10만명당
old    = pd.read_csv('/content/sample_data/old_age.csv')
old=old.drop(columns="동별(1)")
old=old.drop([0], axis=0)
old=old.drop([1], axis=0)
old.rename(columns={
    '동별(2)' : '구',
    '2025 1/4'       : '총인구',      # 실제 파일 이름
    '2025 1/4.1': '고령인구'     # 실제 파일 이름
}, inplace=True)

old

```

```python
for col in ['총인구', '고령인구']:
    old[col] = (old[col].astype(str)
                       .str.replace(',', '') # 1,234 → 1234
                       .str.strip()          # 공백 제거
                       .astype(float))       # 숫자로 변환

# ── 고령화율 계산
old['고령화율'] = old['고령인구'] / old['총인구']
```

```python

df = inst10.merge(old[['구','고령화율']], on='구')
df
```

```python
sns.regplot(data=df, x='고령화율', y='한방기관_10만명당')
r, p = stats.pearsonr(df['고령화율'], df['한방기관_10만명당'])
print(f'상관계수 r={r:.2f}, p={p:.3f}')
```
